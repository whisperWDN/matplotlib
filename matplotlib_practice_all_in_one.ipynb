## plot

Content organized from `plot.py`.

`plt.plot(x, y, ls='-', lw=2, label='plot figure')`

功能：绘制线图，将数据点连接成线。

参数：
- x: x轴数据数组
- y: y轴数据数组
- ls: 线型，如 '-' (实线), '--' (虚线), ':' (点线)
- lw: 线宽，数值越大线越粗
- label: 图例标签，用于legend

```python
x = np.linspace(0.05, 10, 100)
y = np.cos(x)

plt.plot(x, y, ls='-', lw=2, label='plot figure')
plt.legend()
plt.show()
```

## scatter

`plt.scatter(x, y, ls='-', label='scatter figure')`

功能：绘制散点图，在指定位置绘制点。

参数：
- x: x轴数据数组
- y: y轴数据数组
- ls: 线型（但在scatter中通常不适用）
- label: 图例标签

```python
x = np.linspace(0.05, 10, 1000)
y = np.random.randn(1000)

plt.scatter(x, y, ls='-', label='scatter figure')
plt.legend()
plt.show()
```

## xlim

`plt.xlim(0.05, 10)` 和 `plt.ylim(0,1)`

功能：设置x轴和y轴的显示范围。

参数：
- xlim(xmin, xmax): 设置x轴的最小和最大值
- ylim(ymin, ymax): 设置y轴的最小和最大值

```python
x = np.linspace(0.05, 10, 1000)
y = np.random.randn(1000)

plt.scatter(x, y, ls='-', label='scatter figure')
plt.legend()

plt.xlim(0.05, 10)
plt.ylim(0,1)
plt.show()
```

## xlabel

`plt.xlabel('x-axis')` 和 `plt.ylabel('y-axis')`

功能：设置x轴和y轴的标签。

参数：
- xlabel(s): 设置x轴标签，s为字符串
- ylabel(s): 设置y轴标签，s为字符串

```python
x = np.linspace(0.05, 10, 1000)
y = np.sin(x)

plt.plot(x, y, ls='-', lw=2, c = "c", label='plot figure')

plt.legend()

plt.xlabel('x-axis')
plt.ylabel('y-axis')

plt.show()
```

## grid

`plt.grid(linestyle = ":",color = "r")`

功能：添加网格线到图表中。

参数：
- linestyle: 网格线的样式，如 ":" (点线)
- color: 网格线的颜色，如 "r" (红色)

```python
x = np.linspace(0.05, 10, 1000)
y = np.sin(x)

plt.plot(x, y, ls='-', lw=2, c = "c", label='plot figure')

plt.legend()
plt.grid(linestyle = ":",color = "r")

plt.show()
```

## axhline

`plt.axhline(y=0.0,c="r",ls="--",lw=2)` 和 `plt.axvline(x=4.0,c="r",ls="--",lw=2)`

功能：绘制水平参考线(axhline)和垂直参考线(axvline)。

参数：
- axhline(y, c, ls, lw): y为水平线位置，c为颜色，ls为线型，lw为线宽
- axvline(x, c, ls, lw): x为垂直线位置，c为颜色，ls为线型，lw为线宽

```python
x = np.linspace(0.05, 10, 1000)
y = np.sin(x)

plt.plot(x, y, ls='-', lw=2, c = "c", label='plot figure')

plt.legend()

plt.axhline(y=0.0,c="r",ls="--",lw=2)
plt.axvline(x=4.0,c="r",ls="--",lw=2)

plt.show()
```

## axvspan

`plt.axvspan(xmin=4.0, xmax=6.0, facecolor='y', alpha=0.3)` 和 `plt.axhspan(ymin=0.0, ymax=0.5, facecolor='y', alpha=0.3)`

功能：绘制垂直(axvspan)或水平(axhspan)参考区域。

参数：
- axvspan(xmin, xmax, facecolor, alpha): xmin/xmax为x轴范围，facecolor为填充色，alpha为透明度
- axhspan(ymin, ymax, facecolor, alpha): ymin/ymax为y轴范围，facecolor为填充色，alpha为透明度

```python
x = np.linspace(0.05, 10, 1000)
y = np.sin(x)

plt.plot(x, y, ls='-.', lw=2, c = "c", label='plot figure')
plt.legend()

plt.axvspan(xmin=4.0, xmax=6.0, facecolor='y', alpha=0.3)
plt.axhspan(ymin=0.0, ymax=0.5, facecolor='y', alpha=0.3)

plt.show()
```

## annotate

`plt.annotate('maximum', xy=(np.pi/2, 1.0), xytext=((np.pi/2)+1.0, .8),weight='bold',color='b', arrowprops=dict(arrowstyle='->', connectionstyle='arc3', color='b'))`

功能：添加注释文本和箭头到图表中的指定位置。

参数：
- text: 注释文本
- xy: 箭头指向的坐标
- xytext: 文本位置的坐标
- weight: 字体粗细，如 'bold'
- color: 文本颜色
- arrowprops: 箭头属性字典，包括arrowstyle, connectionstyle, color等

```python
x = np.linspace(0.05, 10, 1000)
y = np.sin(x)

plt.plot(x, y, ls='-.', lw=2, c = "c", label='plot figure')
plt.legend()

plt.annotate('maximum', xy=(np.pi/2, 1.0), xytext=((np.pi/2)+1.0, .8),weight='bold',color='b', arrowprops=dict(arrowstyle='->', connectionstyle='arc3', color='b'))

plt.show()
```

## text

`plt.text(3.10, 0.09, 'y-sin(x)', weight='bold', color='b')`

功能：在图表的指定位置添加文本。

参数：
- x, y: 文本位置的坐标
- s: 文本内容
- weight: 字体粗细，如 'bold'
- color: 文本颜色

```python
x = np.linspace(0.05, 10, 1000)
y = np.sin(x)

plt.plot(x, y, ls='-.', lw=2, c = "c", label='plot figure')
plt.legend()

plt.text(3.10, 0.09, 'y-sin(x)', weight='bold', color='b')

plt.show()
```

## title

`plt.title('y=sin(x)')`

功能：设置图表的标题。

参数：
- s: 标题文本字符串

```python
x = np.linspace(0.05, 10, 1000)
y = np.sin(x)

plt.plot(x, y, ls='-.', lw=2, c = "c", label='plot figure')
plt.legend()

plt.title('y=sin(x)')

plt.show()
```

## legend

`plt.legend(loc='upper right')`

功能：显示图例，解释图中的线条或点。

参数：
- loc: 图例位置，如 'upper right' (右上角)

```python
x = np.linspace(0.05, 10, 1000)
y = np.sin(x)

plt.plot(x, y, ls='-.', lw=2, c = "c", label='plot figure')
plt.legend(loc='upper right')

plt.show()
```
