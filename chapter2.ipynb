```python
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np
```

## bar

`plt.bar(x, y, align='center', color='c', tick_label=[...], hatch='/')`

功能：绘制柱状图，用于展示分类数据的数值比较。

参数：
- x: 柱子的位置或分类标签
- y: 柱子的高度值
- align: 柱子对齐方式，如 'center' 或 'edge'
- color: 柱子的填充颜色
- tick_label: x轴刻度标签列表
- hatch: 柱子的填充图案

```python
mpl.rcParams['font.sans-serif'] = ["Simhei"]  # 设置默认字体为 Simhei
mpl.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题

x = [1, 2, 3, 4, 5, 6, 7, 8]
y = [3, 1, 4, 5, 8, 9, 7, 2]

plt.bar(x,y,align='center',color='c',tick_label=['q','a','c','e','r','j','b','q'],hatch='/')

plt.xlabel('箱子编号')
plt.ylabel('箱子重量（kg）')

plt.show()
```

## barh

`plt.barh(x, y, align='center', color='c', tick_label=[...], hatch='/')`

功能：绘制水平柱状图，适合显示类别标签较长或比较值的水平分布。

参数：
- x: 水平条的位置或分类标签
- y: 水平条的长度值
- align: 条形对齐方式
- color: 条形颜色
- tick_label: y轴刻度标签列表
- hatch: 填充图案

```python
mpl.rcParams['font.sans-serif'] = ["Simhei"]  # 设置默认字体为 Simhei
mpl.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题

x = [1, 2, 3, 4, 5, 6, 7, 8]
y = [3, 1, 4, 5, 8, 9, 7, 2]

plt.barh(x,y,align='center',color='c',tick_label=['q','a','c','e','r','j','b','q'],hatch='/')

plt.xlabel('箱子编号')
plt.ylabel('箱子重量(kg)')

plt.show()
```

## hist

`plt.hist(x, bins=bins, color='g', histtype='bar', rwidth=1, alpha=0.6)`

功能：绘制直方图，将连续数据分组后展示频数分布。

参数：
- x: 输入数据数组
- bins: 直方图区间边界或分组个数
- color: 柱子的颜色
- histtype: 直方图类型，如 'bar'
- rwidth: 柱子宽度比例
- alpha: 透明度

```python
mpl.rcParams['font.sans-serif'] = ["Simhei"]  # 设置默认字体为 Simhei
mpl.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题

boxweight = np.random.randint(0, 10, 100)

x = boxweight

bins = np.arange(0, 11, 1)

plt.hist(x, bins=bins, color='g', histtype='bar', rwidth=1, alpha=0.6)
plt.xlabel('箱子重量(kg)')
plt.ylabel('销售数量(个)')

plt.show()
```

## pie

`plt.pie(soldNums, labels=kinds, autopct="%3.1f%%", startangle=60, colors=colors)`

功能：绘制饼图，用百分比分布展示各部分在总量中的占比。

参数：
- x: 各部分的数值
- labels: 部分标签列表
- autopct: 百分比显示格式
- startangle: 起始角度
- colors: 部分颜色列表

```python
mpl.rcParams['font.sans-serif'] = ["Simhei"]  # 设置默认字体为 Simhei
mpl.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题

kinds = ['简易箱', '保温箱', '行李箱', '密封箱']
colors = ['#e41a1c', '#377eb8', '#4daf4a', '#984ea3']

soldNums = [0.05, 0.45, 0.15, 0.35]

plt.pie(soldNums,
        labels=kinds,
        autopct="%3.1f%%",
        startangle=60,
        colors=colors)

plt.title('不同类型箱子的销售数量占比')

plt.show()
```

## plolar

`plt.polar(theta, r, color='chartreuse', linewidth=2, marker='*', mfc='b', ms=10)`

功能：绘制极坐标图，将角度和半径数据绘制在极坐标系中。

参数：
- theta: 角度数据，单位为弧度
- r: 半径数据
- color: 线条颜色
- linewidth: 线宽
- marker: 数据点样式
- mfc: 数据点填充色
- ms: 数据点大小

```python
barSlices = 12

theta = np.linspace(0.0, 2 * np.pi, barSlices, endpoint=False)

r = 30*np.random.rand(barSlices)

plt.polar(theta, r,  color='chartreuse',linewidth=2, marker='*', mfc = "b", ms = 10)

plt.show()
```

## scatter

`plt.scatter(a, b, c=np.random.rand(100), marker='o', s=np.power(10*a+b*20,2), cmap=mpl.cm.RdYlBu)`

功能：绘制散点图，用于展示两个变量之间的关系和点的分布。

参数：
- a, b: x轴和y轴数据
- c: 散点颜色数据
- marker: 散点形状
- s: 散点大小
- cmap: 颜色映射表

```python
a = np.random.rand(100)
b = np.random.rand(100)

plt.scatter(a, b, c=np.random.rand(100), marker='o', s=np.power(10*a+b*20,2), cmap=mpl.cm.RdYlBu)

plt.show()
```

## stem

`plt.stem(x, y, linefmt='-.', markerfmt='o', basefmt='-')`

功能：绘制棒状图与基线组合的图形，一般用于展示离散数据点及其值。

参数：
- x, y: 数据点的位置和值
- linefmt: 茎线样式
- markerfmt: 标记样式
- basefmt: 基线样式

```python
x = np.linspace(0.5, 2*np.pi, 20)
y = np.random.randn(20)

plt.stem(x,y,linefmt='-.', markerfmt='o', basefmt='-')

plt.show()
```

## boxplot

`plt.boxplot(x)`

功能：绘制箱线图，用于展示数据的中位数、四分位数、离群值和分布范围。

参数：
- x: 输入数据数组

```python
mpl.rcParams['font.sans-serif'] = ["FangSong"]  # 设置默认字体为 Simhei
mpl.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题

x = np.random.rand(1000)

plt.boxplot(x)

plt.xticks([1], ['随机数生成器AlpphaRM'])
plt.ylabel('随机数值')
plt.title('随机数生成器抗干扰能力的稳定性')

plt.grid(axis='y',ls=':',alpha=0.4,lw=1,color='gray')

plt.show()
```

## errorbar

`plt.errorbar(x, y, yerr=0.2, fmt='bo:', xerr=0.02)`

功能：绘制带误差条的折线图，展示数据点的不确定性范围。

参数：
- x, y: 数据点坐标
- yerr: y方向误差值
- xerr: x方向误差值
- fmt: 数据点和线条格式

```python
x = np.linspace(0.1, 0.6, 6)
y = np.exp(x)

plt.errorbar(x, y, yerr=0.2, fmt='bo:', xerr=0.02)
plt.xlim(0, 0.7)
plt.show()
```
