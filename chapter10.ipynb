```python
import matplotlib as mpl 
import numpy as np
import matplotlib.pyplot as plt
```

```python
mpl.rc("lines",linewidth=2,color = "c",linestyle="--")
line = {"linewidth":2,"color":"c","linestyle":"--"}
mpl.rc("lines",**line)
```

```python
mpl.rcParams["lines.linewidth"]  = 2
mpl.rcParams["lines.color"]  = "c"
mpl.rcParams["lines.linestyle"]  = "--"
```

```python
plt.axes([0.1,0.7,.3,.3], frameon=True,axisbg="y",aspect="equal")
plt.plot(np.arange(3),[0,1,0])
plt.cla()
plt.plot(np.arange(3),[0,1,0])

plt.axes([0.4,0.4,.3,.3], frameon=True,axisbg="y",aspect="equal")
plt.plot(2+np.arange(3),[0,1,0])

plt.axes([0.7,0.1,.3,.3], frameon=True,axisbg="y",aspect="equal")
plt.plot(4+np.arange(3),[0,1,0])
plt.axis("off")

plt.show()
```
